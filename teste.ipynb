```python
import json

final = json.dumps(recebido, indent=4, ensure_ascii=False)
```

```python
estrutura_final = { 
    "Gestão Estratégica" : 
        [
            {"Bloco" : "GE01", "Pergunta" : "Existem reuniões de acompanhamento?", "Resposta" : "Parcialmente"}
        ]
}
```

```python
estrutura_final = {}

for pilar, data in recebido.items() :
    estrutura_final[pilar] = []

    for bloco, dados in data.items():
        
        for pergunta, resposta in dados.items() :
            estrutura_final[pilar].append({"Bloco" : bloco, "Pergunta" : pergunta, "Resposta" : resposta})
```

```python
for pilar, dados in estrutura_final.items() :
        df = pd.DataFrame(dados)
```

```python
from io import BytesIO

buffer = BytesIO()

with pd.ExcelWriter(buffer, engine='openpyxl') as writer :
    for pilar, dados in estrutura_final.items() :
        df = pd.DataFrame(dados)
        df.to_excel(writer, sheet_name=pilar, index=False )

buffer.seek(0)
```
